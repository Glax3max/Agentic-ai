```python
from langgraph.graph import StateGraph , START,END
from typing import TypedDict
```

```python
class BMIState(TypedDict):

    weight_kg: float
    height_m: float
    bmi:float
```

```python
def calculateBMI(state:BMIState) -> BMIState:
    w = state['weight_kg']
    h = state['height_m']
    bmi = w/(h**2)
    state['bmi'] = round(bmi,2)

    return state
```

```python
graph = StateGraph(BMIState)

#add Nodes to ur graph
graph.add_node('Calculate_BMI', calculateBMI)

#add edges to ur graph
graph.add_edge(START,'Calculate_BMI')
graph.add_edge('Calculate_BMI',END)

#compile the graph
workflow = graph.compile()


```

```python
#execute the graph
intital_state = workflow.invoke({'weight_kg':80,'height_m':2.1})

output_state = workflow.invoke(intital_state)

print(output_state)
```
