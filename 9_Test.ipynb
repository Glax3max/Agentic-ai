```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict  
```

```python
class BugState(TypedDict):
    code:str
    bug:bool
    debug:int
```

```python
def IsBug(state:BugState):
    if(state['bug']) :
     state['debug'] += 1
    
    return state
```

```python
def Debug(state:BugState):
    if(state['debug'] > 0) :
        print("Debug done")
        state['code'] = 'New code'
    return state
```

```python
graph = StateGraph(BugState)

# adding nodes
graph.add_node('Bug',IsBug)
graph.add_node('Debug',Debug)

# adding edges
graph.add_edge(START,'Bug')
graph.add_edge('Bug','Debug')
graph.add_edge('Debug',END)

workflow = graph.compile()
```

```python
intitialState = workflow.invoke({'code':"Sad code",'bug':True,'debug':0})
print(intitialState)
```
