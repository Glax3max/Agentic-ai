```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict
```

```python
class DocumentState(TypedDict):
    citizenship:bool
    age:bool
    education:bool
    workExperience:bool
    
```

```python
def verifyCitizenShip(state:DocumentState)->DocumentState:
    if(state['citizenship'] == True):
        print('Citizenship verified')
    return state
```

```python
def verifyAge(state:DocumentState)->DocumentState:
    if(state['age'] == True):
        print('Age verified')
    return state
```

```python
def verifyEducation(state:DocumentState)->DocumentState:
    if(state['education'] == True):
        print('Education verified')
    return state
```

```python
def verifyWorkExp(state:DocumentState)->DocumentState:
    if(state['workExperience'] == True):
        print('Work Experience verified')
    return state
```

```python
def FullVarificationSummary(state:DocumentState)->DocumentState:
    sum = "The candidate got "
    if(state['citizenship']) :
        sum += 'Citizenship verified '
    else:
        sum +='Citizenship not verified '

    if(state['age']) :
        sum += 'age verified '
    else:
        sum +='age not verified '

    if(state['education']) :
        sum += 'education verified '
    else:
        sum +='education not verified '

    if(state['workExperience']) :
        sum += 'workExp verified.'
    else:
        sum +='workExp not verified.'

    print(sum)
    return state
    
```

```python
graph = StateGraph(DocumentState)

# Adding nodes to the graph
graph.add_node('Citizenship',verifyCitizenShip)
graph.add_node('Age',verifyAge)
graph.add_node('Education',verifyEducation)
graph.add_node('WorkExp',verifyWorkExp)
graph.add_node('Summary',FullVarificationSummary)

# Adding edges to the nodes
graph.add_edge(START,'Citizenship')
graph.add_edge('Citizenship','Age')
graph.add_edge('Age','Education')
graph.add_edge('Education','WorkExp')
graph.add_edge('WorkExp','Summary')
graph.add_edge('Summary',END)

workflow = graph.compile()


```

```python
initial_state = workflow.invoke({'citizenship':False,'age':True,'education':True,'workExperience':True})
print(initial_state)
```
