```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict
```

```python
class TaxState(TypedDict):
    TaxFiled:bool
    TaxAmount:int
    TaxBenifit:int
```

```python
def CheckTaxStatus(state:TaxState)->TaxState :
    if(state['TaxFiled'] == False) :
        print('Tax Not filed')

    return state;
```

```python
def FileTax(state:TaxState)->TaxState:
    if(state['TaxAmount'] > 0):
        state['TaxFiled'] = True
    else:
        print('Not enough money')
    return state
```

```python
def TaxBenifit(state:TaxState) -> TaxState:
    if(state['TaxBenifit'] > 0) :
        state['TaxAmount'] -= state['TaxBenifit']
        print("Got tax benifit")
    return state
```

```python
graph = StateGraph(TaxState)

# Adding nodes to the graph
graph.add_node('TaxStatus',CheckTaxStatus)
graph.add_node('FileTax',FileTax)
graph.add_node('TaxBenifit',TaxBenifit)

# Adding edges to the graph
graph.add_edge(START,)
```
