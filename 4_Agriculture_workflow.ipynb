```python
from  langgraph.graph import StateGraph,START,END
from typing import TypedDict
```

```python
class AgriState(TypedDict) :
    soilReady:bool = False
    seedPlanted:bool = False
    cropGrown:bool = False
    harvested:bool = False
```

```python
def soilReady(state:AgriState)->AgriState:
    state['soilReady'] = True
    print('Soil is ready')
    return state
```

```python
def seedPlanted(state:AgriState)->AgriState:
    state['seedPlanted'] = True
    print('Seed planted')
    return state
```

```python
def cropGrown(state:AgriState)->AgriState:
    state['cropGrown'] = True
    print("Crop grown")
    return state
```

```python
def harvested(state:AgriState)->AgriState:
    state['harvested'] = True
    print('Crop harvested')
    return state
```

```python
graph = StateGraph(AgriState)

# Adding nodes
graph.add_node('Soil',soilReady)
graph.add_node('Seed',seedPlanted)
graph.add_node('Crop',cropGrown)
graph.add_node('Harvest',harvested)

# Adding edge
graph.add_edge(START,'Soil')
graph.add_edge('Soil','Seed')
graph.add_edge('Seed','Crop')
graph.add_edge('Crop','Harvest')
graph.add_edge('Harvest',END)

workflow = graph.compile()
```

```python
initial_state = workflow.invoke({})
final_state = workflow.invoke(initial_state)
print(final_state)
```
