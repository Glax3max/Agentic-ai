```python
from  langgraph.graph import StateGraph,START,END
from typing import TypedDict
```

```python
class AgriState(TypedDict) :
    soilReady:bool
    seedPlanted:bool
    cropGrown:bool
    harvested:bool 
```

```python
def soilReady(state:AgriState)->AgriState:
    if(state['soilReady']):
        print('Soil is ready')
    else:
        print('Soil is not ready')
    return state
```

```python
def seedPlanted(state:AgriState)->AgriState:
    if(state['seedPlanted']):
        print('Seed planted')
    else:
        print('Seed is not Planted')
    return state
```

```python
def cropGrown(state:AgriState)->AgriState:
    if(state['cropGrown']):
        print("Crop grown")
    else:
        print('Crop is not grown')
    return state
```

```python
def harvested(state:AgriState)->AgriState:
    if(state['harvested']):
        print('Crop harvested')
    else:
        print('Crop not harvested')
    return state
```

```python
graph = StateGraph(AgriState)

# Adding nodes
graph.add_node('Soil',soilReady)
graph.add_node('Seed',seedPlanted)
graph.add_node('Crop',cropGrown)
graph.add_node('Harvest',harvested)

# Adding edge
graph.add_edge(START,'Soil')
graph.add_edge('Soil','Seed')
graph.add_edge('Seed','Crop')
graph.add_edge('Crop','Harvest')
graph.add_edge('Harvest',END)

workflow = graph.compile()
```

```python
initial_state = workflow.invoke({'soilReady':True,'seedPlanted':True,'cropGrown':False,'harvested':True})
# final_state = workflow.invoke(initial_state)
print(initial_state)
```
