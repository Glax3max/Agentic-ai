```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict
```

```python
class BatsmanState(TypedDict):

    runs:int
    balls:int
    fours:int
    sixs:int

    sr:float
    bpb:float
    boundry_percent:float
    summary:str
```

```python
def calculate_sr(state:BatsmanState):
    runs = state['runs']
    balls = state['balls']

    sr = runs/balls
    state['sr'] = sr
    return {'sr':sr}
 
```

```python
def calculate_bpb(state:BatsmanState):
    balls = state['balls']
    boundary = state['fours'] + state['sixs']

    state['bpb']  = balls/boundary
    return {'bpb':balls/boundary}
```

```python
def calculate_boundary_percent(state:BatsmanState):
    boundary = state['fours'] + state['sixs']
    balls = state['balls']

    boundary_percent = (boundary/balls)*100
    state['boundry_percent'] = boundary_percent
    return {'boundry_percent':boundary_percent}
```

```python
def summary(state:BatsmanState):
    summary = f'So we got the SR: {state['sr']} and BPB: {state['bpb']} and Boundary_Percent: {state['boundry_percent']}'
    print(summary)
    return {'summmary':summary}
```

```python
graph = StateGraph(BatsmanState)

# adding node
graph.add_node('calculate_sr',calculate_sr)
graph.add_node('calculate_bpb',calculate_bpb)
graph.add_node('calculate_boundary_percent',calculate_boundary_percent)
graph.add_node('summary',summary) 

# adding edges between the nodes
graph.add_edge(START,'calculate_sr')
graph.add_edge(START,'calculate_bpb')
graph.add_edge(START,'calculate_boundary_percent')
graph.add_edge('calculate_sr','summary')
graph.add_edge('calculate_bpb','summary')
graph.add_edge('calculate_boundary_percent','summary')
graph.add_edge('summary',END)

workflow = graph.compile()

```

```python
initialState  = {'runs':129,'balls':43,'fours':8,'sixs':6}
final_State = workflow.invoke(initialState)
# print(final_State)
workflow
```
